```python
import pandas as pd

df = pd.read_csv(r'C:\Users\Eiabe\Desktop\venda_loja\vendas_loja_nivel2.csv')
```

```python
df.head()
```

```python
df = df.dropna(subset=['quantidade'])
df['cliente'] = df['cliente'].fillna('Desconhecido')

df['quantidade'] = df['quantidade'].astype(int)
df['data_venda'] = pd.to_datetime(df['data_venda'])

duplicatas = ['produto', 'categoria', 'cliente', 'quantidade', 'preco_unitario', 'data_venda']
df = df[duplicatas].drop_duplicates()

df['valor_total'] = df['quantidade'] * df['preco_unitario']

padronizacao = ['produto', 'categoria', 'cliente']

for i in padronizacao:

    df[i] = df[i].str.title()
    df[i] = df[i].str.strip()

faturamento_total = (df['quantidade'] * df['preco_unitario']).sum()
categoria_valor = df.groupby('categoria')['valor_total'].sum()

quantidade_vendidos = df.groupby('produto').agg({
    'quantidade': 'sum',
    'valor_total': 'sum'
}).reset_index()

categoria_valor = categoria_valor.reset_index()

quantidade_vendidos = quantidade_vendidos.sort_values(by = 'quantidade', ascending = False)
categoria_valor = categoria_valor.sort_values(by = 'valor_total', ascending = False)

df.info()
```
